#### Explore Dataset


### 1. Load All Sheets from Excel

```python
# Requirement Libraries 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

excel_path = '../data/raw/my_data.xlsx'
all_sheets = pd.read_excel(excel_path, sheet_name=None)
```

### 2. Preview Data

```python
# Check the first few rows from each sheet
for sheet, df in all_sheets.items():
    print(f"Sheet: {sheet}")
    display(df.head())
```

```python
domain_map = {
    'Sheet_1': 'Culture/Society',
    'Sheet_2': 'History/Biography',
    'Sheet_3': 'Sports',
    'Sheet_4': 'Economy/Energy',
    'Sheet_5': 'Health/Government',
    'Sheet_6': 'Technology/Business',
    'Sheet_7': 'Science/Personal',
    'Sheet_8': 'Media/Education'
}
```

```python
#Visualize domain distribution
domain_counts = {domain_map[sheet]: len(df) for sheet, df in all_sheets.items()}
plt.figure(figsize=(8,5))
plt.bar(domain_counts.keys(), domain_counts.values())
plt.title('Sentence Count per Domain')
plt.xlabel('Domain')
plt.ylabel('Sentence Count')
plt.xticks(rotation=45)
plt.show()
```

### 3. Check for Missing Values

```python
for sheet, df in all_sheets.items():
    print(f"Sheet: {sheet} - Missing values: {df.isnull().sum().sum()}")
```

```python
#  Missing Values per Sheet
missing_counts = {sheet: df.isnull().sum().sum() for sheet, df in all_sheets.items()}

plt.figure(figsize=(10,6))
sns.barplot(x=list(missing_counts.keys()), y=list(missing_counts.values()))
plt.title('Missing Values per Sheet')
plt.xlabel('Sheet Name')
plt.ylabel('Missing Value Count')
plt.xticks(rotation=45)
plt.show()
```

### 4. Check Sentence Lengths

```python
for sheet, df in all_sheets.items():
    lengths = df.iloc[:,0].astype(str).apply(lambda x: len(x.split()))
    print(f"Sheet: {sheet} - Min: {lengths.min()}, Max: {lengths.max()}, Mean: {lengths.mean()}")
```

```python
# Visualize sentence length distribution for all sheets
all_lengths = []
for sheet, df in all_sheets.items():
    lengths = df.iloc[:,0].astype(str).apply(lambda x: len(x.split()))
    all_lengths.extend(lengths)

plt.figure(figsize=(10,6))
sns.histplot(all_lengths, bins=30, kde=True)
plt.title('Sentence Length Distribution (All Sheets)')
plt.xlabel('Number of Words')
plt.ylabel('Frequency')
plt.show()
```

### 5. Look for Duplicates

```python
for sheet, df in all_sheets.items():
    print(f"Sheet: {sheet} - Duplicates: {df.duplicated().sum()}")
```

```python

duplicate_counts = {sheet: df.duplicated().sum() for sheet, df in all_sheets.items()}

plt.figure(figsize=(8,5))
sns.barplot(x=list(duplicate_counts.keys()), y=list(duplicate_counts.values()))
plt.title('Duplicate Sentences per Sheet')
plt.xlabel('Sheet Name')
plt.ylabel('Duplicate Count')
plt.xticks(rotation=45)
plt.show()
```

### 6. Spot Unwanted Characters

```python
# Check for non-Uzbek characters, numbers, or symbols:
import re

for sheet, df in all_sheets.items():
    sample = df.iloc[:,0].astype(str).sample(10, random_state=1)
    for text in sample:
        if re.search(r'[^a-zA-Zа-яА-ЯёЁўқғҳЎҚҒҲʼ‘’“”.,!?0-9\s-]', text):
            print(f"Unusual chars in: {text}")
```

```python
# Show Example Problematic Sentences
problematic = []
for sheet, df in all_sheets.items():
    sample = df.iloc[:,0].astype(str)
    for text in sample:
        if re.search(r'[^a-zA-Zа-яА-ЯёЁўқғҳЎҚҒҲʼ‘’“”.,!?0-9\s-]', text):
            problematic.append({'sheet': sheet, 'text': text})
problematic_df = pd.DataFrame(problematic)
display(problematic_df.head(10))
```
